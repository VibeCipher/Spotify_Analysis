```python
if 'release_date' in df_tracks.columns:
    df_tracks['release_date'] = pd.to_datetime(df_tracks['release_date'], errors='coerce')
    df_tracks.set_index('release_date', inplace=True)
elif df_tracks.index.name == 'release_date':
    df_tracks.index = pd.to_datetime(df_tracks.index, errors='coerce')
else:
    print("release_date not found in columns or index.")
df_tracks.head(3)
```

```python
df_tracks.columns = df_tracks.columns.str.strip().str.lower()

if 'duration_ms' in df_tracks.columns:
    df_tracks['duration'] = df_tracks['duration_ms'].apply(lambda x: round(x / 1000))
    df_tracks.drop('duration_ms', axis=1, inplace=True)
else:
    print("'duration_ms' column not found.")
```

```python
df_tracks.duration.head(3)
```

```python
corr_df = df_tracks.select_dtypes(include=['number']).corr(method='pearson')
plt.figure(figsize=(15, 7))
heatmap = sns.heatmap(corr_df, annot=True, fmt=".1g", vmin=-1, vmax=1, center=0, cmap="rocket", linewidths=1, linecolor="Green")
heatmap.set_title("Correlation heatmap between variables")
heatmap.set_xticklabels(heatmap.get_xticklabels(), rotation=80)
```

```python
df_tracks['dates'] = df_tracks.index.get_level_values('release_date')
df_tracks['dates'] = pd.to_datetime(df_tracks['dates'], format='mixed', errors='coerce')
years = df_tracks['dates'].dt.year
```

```python
sns.displot(years,discrete=True,aspect=2,height=6,kind="hist").set(title="Number of song per year")
```

```python
duplicates = df_tracks.index.duplicated()
print(df_tracks[duplicates])
```

```python
df_tracks['year'] = df_tracks.index.year
```

```python
df_yearly = df_tracks.groupby('year')['duration'].mean().reset_index()
fig_dim = (18, 7)
fig, ax = plt.subplots(figsize=fig_dim)
sns.barplot(
    x='year',
    y='duration',
    data=df_yearly,
    ax=ax,
    hue='year',  
    palette='magma', 
    err_kws={'linewidth': 0},
    legend=False 
)
ax.set_title("Years vs Average Track Duration (in seconds)", fontsize=16, fontweight='bold')
ax.set_xlabel("Year", fontsize=12)
ax.set_ylabel("Average Duration (s)", fontsize=12)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
df_tracks['year'] = df_tracks.index.year 
df_yearly = df_tracks.groupby('year')['duration'].sum().reset_index()
sns.set_style("whitegrid")
fig_dims = (10, 5)
fig, ax = plt.subplots(figsize=fig_dims)
sns.lineplot(x='year', y='duration', data=df_yearly, ax=ax)
ax.set_title("Years vs Total Duration", fontsize=16, fontweight='bold')
plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
df_genre = pd.read_csv("SpotifyFeatures.csv")
```

```python
df_genre.head(5)
```

```python
plt.title("Duration of the Song for Different Genres", fontsize=16, fontweight='bold')
sns.set_style("whitegrid")
sns.barplot(y='genre', x='duration_ms', data=df_genre, hue='genre', palette="inferno", err_kws={'linewidth': 0}, legend=False)
plt.xlabel("Duration in Milliseconds", fontsize=12)
plt.ylabel("Genres", fontsize=12)
plt.tight_layout()
plt.show()
```

```python
sns.set_style("darkgrid")
plt.figure(figsize=(10, 5))
famous = df_genre.sort_values("popularity", ascending=False).head(10)
sns.barplot(
    y='genre',
    x='popularity',
    data=famous,
    hue='genre',
    palette='Set2',     
    legend=False      
).set(title="Top 10 Genres by Popularity")
plt.xlabel("Popularity")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```
